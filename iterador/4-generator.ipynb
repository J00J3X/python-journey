Yield
- Is a keyword used to create generators. When you use it in a function, you transform the function into a generator, which is a special type of iterator
- Yield: can be called again and resume the execution at the point where it paused, allowing the production of a sequence of values over time

```python
def odd_numbers(n):
    for i in range(1, n+1, 2):
        yield i
```

```python
for number in odd_numbers(10):
    print(number)
```
