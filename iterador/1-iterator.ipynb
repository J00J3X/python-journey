```python
class Counter:
    def __init__(self, limit):
        self.limit  = limit
        self.actual = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.actual > self.limit:
            raise StopIteration
        else:
            value = self.actual
            self.actual += 1 
            return value
```

```python
counter = Counter(10)
for number in counter:
    print(number)
```
