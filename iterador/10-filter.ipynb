Filter
- Is used to filter elements of an iterable based on a function true or false.
- The objective is allow only elements that satisfies a condition.

```python
def is_pair(x):
    return x % 2 == 0
```

```python
numbers = [1,2,3,5,6,8,987,654,3,2,]
```

```python
pair_numbers = filter(is_pair, numbers)
list(pair_numbers)
```

```python
products = [
{"name":"Product A", "price": 10.0, "in_stock": True},
{"name":"Product B", "price": 70.0, "in_stock": False},
{"name":"Product C", "price": 80.0, "in_stock": False},
{"name":"Product D", "price": 1.0, "in_stock": True},
]
```

```python
def product_in_stock(product):
    return product["in_stock"]
```

```python
products_avaliable = filter(product_in_stock, products)
list(products_avaliable)
```

```python
transactions = [200.0, -50.0, 75.0, -30.0, 190.0]
```

```python
def is_expense(transaction):
    return transaction < 0
```

```python
revenues = filter(is_expense, transactions)
list(revenues)
```
