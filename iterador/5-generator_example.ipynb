```python
def lines_with_number_of_words(arqchive, number):
    with open(arqchive, 'r', encoding='utf-8') as file:
        for line in file:
            if len(line.split()) == number:
                yield line.strip()
```

```python
archive = 'text.txt'
number_words = 3
for line in lines_with_number_of_words(archive, number_words):
    print(line)
```
