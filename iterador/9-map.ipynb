Map
- Function that applies a function to every iterable item(like a list, for example)
- Is an efficient way and concise to transform data

```python
def square(x):
    return x ** 2
```

```python
numbers = [1,2,3,4,5]
```

```python
for n in numbers:
    print(square(n))
```

```python
results = map(square, numbers)
list(results)
```

```python
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32
```

```python
temperatures_celsius = [0,20,37,100]
```

```python
temperatures_fahrenheit = map(celsius_to_fahrenheit, temperatures_celsius)
list(temperatures_fahrenheit)
```

```python
def normalizate_name(name):
    return name.strip().lower()
```

```python
names = ["    Anna", "    jacsok", "ana", "Banana   ", "MARELO  " ]
```

```python
normalized_names = map(normalizate_name, names)
list(normalized_names)
```
