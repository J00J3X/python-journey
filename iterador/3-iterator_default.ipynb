```python
import itertools
```

```python
# Example 1 - Range()
for i in range(5):
    print(i)
```

```python
# Example 2 - Iterate List
fruits = ['Apple', 'Banana', 'Orange']
for fruit in fruits:
    print(fruit)
```

```python
# Example 3 - Iterate dictionary
person = {'name':'Alice', 'age': 30}
for key in person:
    print(key, person[key])
```

```python
# Example 4 - Enumerate()
for index, fruit in enumerate(fruits):
    print(index, fruit)
```

```python
# Example 5 - Zip()
names   = ['Alice', 'Bob', 'Charles']
ages    = [30, 20, 50]
for name, age in zip(names, ages):
    print(name, ages)
```

```python
# Example 6 - Itertools
for i in itertools.count(start=10, step=2):
    if i > 20:
        break
    print(i)
```
