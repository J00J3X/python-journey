```python
#1 - Generator Expression
squares = (x ** 2 for x in range(10))
for square in squares:
    print(square)
```

```python
#2 - Infinite Iterator
def counter_infinite():
    i = 0
    while True:
        yield 1
        i += 1
counter = counter_infinite()
for _ in range(10):
    print(next(counter))
```
