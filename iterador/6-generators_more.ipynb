```python
#1 - Generator Expression
squares = (x ** 2 for x in range(10))
for square in squares:
    print(square)
```

```python
#2 - Infinite Iterator
def counter_infinite():
    i = 0
    while True:
        yield i
        i += 1
counter = counter_infinite()
for _ in range(10):
    print(next(counter))
```

```python
#3 - Value collector with send()
def generator_sum():
    total = 0
    while True:
        value = yield total # Get a value and return an accumulated total
        if value is not None:
            total += value

total_gen = generator_sum()
next(total_gen)
print(total_gen.send(10))
print(total_gen.send(10))
print(total_gen.send(10))
```

```python
# 4 - Exception treatment
def generator_exception():
    try:
        while True:
            value = yield
            print(f"Value recived: {value}")
    except ValueError:
        print("Invalid value!")

g = generator_exception()
next(g)
g.send(10)
g.throw(ValueError)
```

```python
# 5 - Chained generators 
def multiplicate_by_two(iterable):
    for i in iterable:
        yield i * 2

def add_five(iterable):
    for i in iterable:
        yield i + 5

numbers = range(5)
result  = add_five(multiplicate_by_two(numbers))
for r in result:
    print(r)
    
```
