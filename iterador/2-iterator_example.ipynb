```python
class NameWithLetter:
    def __init__(self, names, letter):
        self.names  = names
        self.letter = letter
        self.index  = 0

    def __iter__(self):
        return self

    def __next__(self):
        while self.index < len(self.names):
            name = self.names[self.index]
            self.index += 1
            if name.startswith(self.letter):
                return name 
        raise StopIteration
```

```python
name_list = ['Jonathan', 'Krefta', 'Jonas', 'James']
letter = 'J'
iterator_names = NameWithLetter(name_list, letter)

for name in iterator_names:
    print(name)
```
