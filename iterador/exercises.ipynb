```python
import re
```

```python
def extract_urls(text):
    url_pattern = r'https?://[a-zA-Z0-9.-]+\.[a-zA-Z]{2,}'
    urls        = re.findall(url_pattern, text)
    return urls
```

```python
example_text = """
Visite nosso site em https://www.exemplo.com e 
https://blog.exemplo.org 
"""

urls_founded = extract_urls(example_text)
print(urls_founded)
```

```python
def validate_password(password):
    password_patten = r'^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)[A-Za-z\d]{8,}$'
    return bool(re.match(password_patten, password))
```

```python
passwords = ["Password123", "Word5002", "kaswwords33", "AAAAASWORD34", ]
for p in passwords:
    print(f"The password {p} is valid? {validate_password(p)}")
```

```python
def words_counter(text):
    word_pattern    = r'\b\w+\b'
    words           = re.findall(word_pattern, text)
    return len(words)
```

```python
example_text    = "Hello world! This is a example phrase to count words"
wordsCounter   = words_counter(example_text)
print(wordsCounter)
```
